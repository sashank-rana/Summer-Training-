```python
batsman = pd.read_csv("/content/sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.title("V Kohli vs RG Sharma")
plt.xlabel("Season")
plt.ylabel("Runs")
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color = "#03fce3", linestyle = "dashdot", linewidth = 2, marker="D", markersize=5, label="VK")
plt.plot(batsman['index'], batsman['RG Sharma'], color = "#fc0303", linestyle = "dotted", linewidth = 2, marker="o", markersize=5,  label="RS")
plt.title("V Kohli vs RG Sharma")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.legend()
plt.grid()
```
