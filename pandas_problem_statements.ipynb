### ***Question 1--> Student Performance Analyzer***

```python
import pandas as pd
import numpy as np
data = pd.read_csv("/content/StudentMarksDataset.csv")
data
```

```python
data["Average"] = data[["Std_StudyHours", "Std_Marks"]].mean(axis=1)
data
```

```python
top_10 = data.sort_values(by="Std_Marks", ascending=False).head(10)
top_10
```

```python
highest_per_branch = data.groupby('Std_Branch')['Std_Marks'].max().reset_index()
highest_per_branch
```

```python
def assign_grade(marks):
    if marks >= 85:
        return 'A'
    elif marks >= 75:
        return 'B'
    elif marks >= 65:
        return 'C'
    elif marks >= 50:
        return 'D'
    else:
        return 'F'
data['Grade'] = data['Std_Marks'].apply(assign_grade)
data
```

```python
data.describe()
```

```python
import pandas as pd
data = pd.read_csv("/content/archive (3).zip")
data
```

```python
dept_avg = data.groupby("experience")["income"].mean()
dept_avgdept_avg = data.groupby("experience")["income"].mean()
dept_avg
```

```python
print("=== INCOME BY EXPERIENCE===")
for k, v in data.groupby(pd.cut(data['experience'], bins=[-1, 3, 7, 11, 15]), observed=False)['income'].mean().items():
    print(f"{str(k):<18} | {'█' * int(v/2000)} ${v:,.0f}")
print("\n=== INCOME BY AGE GROUP ===")
for k, v in data.groupby(pd.cut(data['age'], bins=[20, 30, 40, 50, 60]), observed=False)['income'].mean().items():
    print(f"{str(k):<18} | {'█' * int(v/2000)} ${v:,.0f}")
```
